```python
import pandas as pd

# Step 1: Create sample student data
data = {
    'Name': ['Aman', 'Rahul', 'Priya', 'Sneha', 'Rahul', 'Karan'],
    'Marks': [85, 90, None, 78, 90, None],
    'City': ['Pune', 'Mumbai', 'Delhi', None, 'Mumbai', 'Pune']
}

df = pd.DataFrame(data)

# Step 2: Save data to CSV file
df.to_csv('students.csv', index=False)

# Step 3: Read CSV file
df = pd.read_csv('students.csv')

print("Original Data:")
print(df)

# Step 4: Explore the dataset
print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

# Step 5: Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 6: Fill missing values
df['Marks'] = df['Marks'].fillna(df['Marks'].mean())
df['City'] = df['City'].fillna(df['City'].mode()[0])

# Step 7: Check duplicate records
print("\nDuplicate Records:")
print(df.duplicated())

# Step 8: Remove duplicate records
df = df.drop_duplicates()

# Step 9: Select data using loc[]
print("\nStudents with Marks above 80:")
print(df.loc[df['Marks'] > 80])

# Step 10: Remove unnecessary column
df = df.drop(columns=['City'])

# Step 11: Display cleaned data
print("\nCleaned Data:")
print(df)

# Step 12: Save cleaned data
df.to_csv('cleaned_students.csv', index=False)

print("\nData Preprocessing Completed!")
```
